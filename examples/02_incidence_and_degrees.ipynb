# 02 关联矩阵与度序列

本 notebook 演示两个最常用的度量工具：

- `honet.incidence_matrix(hg)`：关联矩阵 `B`，形状 `结点数 × 超边数`，`B[i][j] = 1` 表示第 `i` 个结点属于第 `j` 条超边；
- `honet.degree_sequence(hg)`：高阶度序列 `k_i = |{e : i ∈ e}|`，按降序返回。

**预期看到**：4×3 的 0/1 矩阵、每列和等于超边大小、度序列 `[2, 2, 2, 1]`。

矩阵行列顺序始终与 `hg.nodes`、`hg.edges` 一致，因此可以直接对上下标。

```python
import honet

hg = honet.Hypergraph(
    nodes=["a", "b", "c", "d"],
    edges=[
        ["a", "b", "c"],
        ["a", "b"],
        ["c", "d"],
    ],
)

matrix = honet.incidence_matrix(hg)

# 打印带表头的文本表格，便于肉眼核对
header = "      " + "  ".join(f"e{j}" for j in range(hg.num_edges))
print(header)
for node, row in zip(hg.nodes, matrix, strict=True):
    print(f"{node:>4}  " + "   ".join(str(v) for v in row))
```

## 用矩阵验证结构

- 每列之和应等于对应超边的成员个数；
- 每行之和应等于该结点的高阶度，也就是度序列里的一个值。

```python
col_sums = [sum(row[j] for row in matrix) for j in range(hg.num_edges)]
row_sums = [sum(row) for row in matrix]

print("各列和（=超边大小）:", col_sums, "| edge_sizes:", list(hg.edge_sizes))
print("各行和（=高阶度）  :", row_sums)
print("度序列（降序）    :", honet.degree_sequence(hg))
```

## 度分布

`degree_sequence` 只返回排序后的列表；做度分布统计时可以直接在其上计数。
下面是“度 → 具有该度的结点数”的简单直方图。

```python
from collections import Counter

degrees = honet.degree_sequence(hg)
dist = Counter(degrees)

for k in sorted(dist, reverse=True):
    print(f"度 {k}: {dist[k]} 个结点  " + "#" * dist[k])

print()
print(f"结点数={hg.num_nodes} 超边数={hg.num_edges} 平均度={sum(degrees) / len(degrees):.2f}")
```
